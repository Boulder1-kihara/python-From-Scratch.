## **Function Arguments**

Arguments are values passed to a function when it is called.

They allow functions to receive input data and perform operations using those values.

```python
def evenodd(x):
  if(x % 2 == 0):
    return "even"
  else:
    return "odd"

print(evenodd(16))
print(evenodd(7))
```

## **Types of function arguments**

1. Default arguments: They use a pre-defined value when no value is passed during function call.

```python
def my_function(x, y = 50):
  print("x: ", x)
  print("y: ", y)

my_function(10)
```

2. keyword arguments: pass values using parameter names, so argument order does not matter.

```python
def student(fname, lname):
  print(fname, lname)

student(fname = "Abel", lname = "Kihara")
student(lname = "Kihara", fname = "Abel")
```

3. Positional Arguments: Values are assigned to parameters based on their order in the function call

```python
def user(name, age):
  print("my name is ", name)
  print(f"i am {age} years old")

user("Abel", 20)
user(27, "Abel")
```

## **Python Pass Statement**

This is a python placeholder that does nothing when exwcuted.



*   It is used to keep code blocks valid where a statement is required but no logic is needed yet.

*   Examples situations where pass is used are empty functions, classes, loops or conditional blocks.



### **In functions**

The pass keyword in a finction is used when er define a function but deont want to implement its logic immediately.

it allowes the function to be syntactically valid, even though it does not perform any actions yet.

```python
def fun():
  pass

fun()
```

## **In Conditional Statements**

In conditionak statements when no action is needed but a block is still required, pass statement acts as a placeholder to keep the code syntactically valid.

```python
x = 10

if x > 5:
    pass

else:
  print("x is 5 or less")
```

## **in loops**

in loops, pass can be used to skip writing any action during a specific iteration while still keeping the loop structure correct

```python
for i in range(5):
  if i == 3:
    pass
  else:
    print(i)
```

## **In classes**

The pass statement allows defining empty classes or methods that act as placeholders until actual functionality is added later

```python
class empty_class:
  pass

class person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def greet(self):
    pass

p = person("john", 30)
```

## **Local Variables**

Local variables are defined inside a function and exist only during its execution.

They cannot be accessed from outside the function.

```python
def greet():
  msg = "Hello Squad"
  print(msg)

greet()
```

```python
#this will result to an error because printing it outside the function does not exist
def greetings():
  mesg = "hello"
  print("inside function:", mesg)

greet()
print("outside function:", mesg)
```

## **Global Variables**

Global variables are declared outside the function and can be accessed anywhere in the program including inside functions.

```python
msg = "hello squad"

def messge():
  print(msg)

messge()
print("The message is:", msg)
```

## **Use of Local and Global Variables**

if a variable is defined both globally and locally with the same name, local variable shadowsthe global variable inside the function.

```python
def fun():
  s = "hello student"
  print(s)

s = "This is python"
fun()
print(s)
```

By default, one cannot modify a global variable inside a function without declaring it as global. if not it will generate an error because it treats variable as local.

To modify it, we use the global keyword

```python
def funs():
  ss += 'gfg'
  print(ss)

ss = "This is python"
funs()
```

# **Python String**

Strings are a sequence of characters inside quotes. It can include letters, numbers, symbols and spaces.

Strings are mainly used for text handling and manipulation.

## **Creating a python String**
Strings  can be created using either single(' ') or double(" ") quotes, but both are the same.

```python
a = "hello"
b = 'squad'

print(f"{a} {b}")
```

### **Multi-line Strings**
Multiple line strings are creaed using single triple(''' ''') or double-triple(""" """). Newlines are preserved.

```python
a = """hello
my name is Abel"""

b = '''and I'm a student'''

print(f"{a}, {b}")
```

## **Accessing Characters in String**
Strings are indexed sequences. Positive indices start at 0 from the left, while negative indices start at -1 from the right as represented.

```python
#print the element at the index
s = "abcdef"
print(s[0])
print(s[4])
```

```python
#Read character from the end
s = 'abcdef'

print(s[-3])
print(s[-4])
```

## **String Slicing**
Slicing is a way of extracting a string by specifying the start and the end indices.

The syntax for slicing is string [start:end] where start starting index and end is stoping index

```python
s = "abcdef"
print(s[1:3])
```

```python
s = "abcdef"
print(s[2:5])
```

## **Looping Through Strings**
Strings are iterable meaning we can access each character one by one ising a loop

```python
s = "abcde"
for i in s:
  print(i)
```

## **Deleting a String**
Individual characters in a string cann not be deleted since strings are immutable.

To delete an entire string, we use the del keyword

```python
s = "abcdef"
del s
print(s)
```

## **Updating a String**
Srtrings cannot be changed directly after creation, So modification results to a new string being created using slicing or methods like `replace()`.

```python
s = "abcde fgh"
s1 = "i" + s[1:]
s2 = s.replace("ABC", "abc")
s3 = "ibcde fgh"

print(s1)
print(s2)
print(s3)
```

## **Common String Methods**
there are common  string methods auch as

1. **len()**: It returns the no.of characters in a string including spaces and punctuation.

2. **upper() and lower()**: Upper converts all characters to upper case while lower converts all cases to lower case.

```python
#1. len method
a = "abcdef"
print(len(a))
```

```python
#2. upper and lower
a = "Abcdef"
print(a.upper())
print(a.lower())
```

3. **strip() and replace()**: strip() removes leading and trailing whitespace from the string while replace() replaces all occurences of a specified substring with another

```python
s = " abcdef "
print(s.strip())

s = "python is good"
print(s.replace("good", "awesomr"))
```
